# Refactoring prepare data in function

```python
import pandas as pd
```

```python
def _is_x_true(x:pd.Series) -> pd.Series:
    return x == "t"

def _parse_percentage(x:pd.Series) -> pd.Series:
    return x.str.replace("%", "").astype(float)

def _parse_money(x:pd.Series) -> pd.Series:
    return x.str.replace("$", "").str.replace(",", "").astype(float)
```

```python
# test functions
companies = catalog.load("companies")
print("companies.head(): \n", companies.head().to_string(), "\n")
print("Treat the 'iata_approved' and 'company_rating' columns")
companies["iata_approved"] = _is_x_true(companies["iata_approved"])
companies["company_rating"] = _parse_percentage(companies["company_rating"])

print("companies.head(): \n", companies.head().to_string(), "\n")

shuttles = catalog.load("shuttles")
print("shuttles.head(): \n", shuttles.head().to_string(), "\n")
print("Treat the 'd_check_complete', 'moon_clearance_complete' and 'price' columns")
shuttles["d_check_complete"] = _is_x_true(shuttles["d_check_complete"])
shuttles["moon_clearance_complete"] = _is_x_true(shuttles["moon_clearance_complete"])
shuttles["price"] = _parse_money(shuttles["price"])
print("shuttles.head(): \n", shuttles.head().to_string(), "\n")
```
